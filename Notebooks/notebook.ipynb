| Column     | Description              |
|------------|--------------------------|
| `'timestamp'` | Date de la transaction. |
| `'time_since_login_min'` | Temps écoulé depuis la connexion de l’utilisateur à l’application. |
| `'transaction_amount'` | Montant, en livres sterling (£), envoyé vers un autre compte. |
| `'transaction_type'` | Type de transaction : <ul><li>`CASH-OUT` – Retrait d’argent d’un compte.</li><li>`PAYMENT` – Paiement vers un tiers.</li><li>`CASH-IN` – Opération inverse d’un retrait : dépôt d’argent sur un compte.</li><li>`TRANSFER` – Virement de fonds d’un compte à un autre.</li> |
| `'is_first_transaction'` | Indicateur binaire précisant s’il s’agit de la première transaction de l’utilisateur (1 pour une première transaction, 0 sinon). |
| `'user_tenure_months'` | Ancienneté en mois depuis la création du compte de l’utilisateur ou depuis son inscription. |
| `'is_fraud'` | Étiquette binaire indiquant si la transaction est frauduleuse (1 pour fraude, 0 sinon). |
| `'predicted_fraud_proba'` | Probabilité attribuée par un modèle de détection indiquant la vraisemblance qu’une transaction soit frauduleuse. |
| `'predicted_fraud'` | Étiquette de classification prédite, calculée à partir de la probabilité de fraude estimée par le modèle de détection (1 pour fraude prédite, 0 sinon). |
